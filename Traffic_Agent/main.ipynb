Traffic Environment

```python
traffic ={}#empty dictionary

roads=['North','South','East','West']

for road in roads:
    vehicles=int(input(f"Enter vehicles in {road}: '"))

    traffic[road] = vehicles

    print("\nCurrent Traffic Condition: ")
    for road,vehicles in traffic.items():
        print(road,":",vehicles)
```

```python
#Add Agent
traffic ={}
roads=['North','South','East','West']

for road in roads:
    vehicles= int(input(f"Enter the number of vehicles in {road}: "))
    traffic[road]= vehicles

    #Intelligent Agent

    def trafficagent(traffic):
        selected_road=max(traffic,key=traffic.get)
        return selected_road

# Traffic Light COntroller

def control_traffic_light(selected_road):
    for road in roads:
        if road == selected_road:
            print(road,": Green")
        else:
            print(road,": Red")

#Run agent
decision = trafficagent(traffic)
print("\nSelected Road:", decision)
control_traffic_light(decision)            
```
